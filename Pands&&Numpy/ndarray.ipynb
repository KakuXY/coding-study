```python
import numpy as np
```

# ndarrayの特性

```python
arr = np.array(5)
print(arr)
print(arr.ndim)
```

```python
arr = np.array([1, 2, 3])
print(arr)
print(arr.ndim)
```

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr)
print(arr.ndim)
```

```python
arr = np.array([[[1,2],[2,3]], [[3,4], [4, 5]]])
print(arr)
print(arr.ndim)
```

# ndarrayの属性

```python
arr = np.array(1)
print(arr)
print(arr.shape)
print(arr.ndim)
print(arr.size)
print(arr.dtype)
print(arr.T)
```

```python
arr = np.array([1, 2, 3])
print(arr)
print(arr.shape)
print(arr.ndim)
print(arr.size)
print(arr.dtype)
print(arr.T)
```

```python
arr = np.array([[1,2,3], [4, 5, 6]])
print(arr)
print(arr.shape)
print(arr.ndim)
print(arr.size)
print(arr.dtype)
print(arr.T)
```

# ndarrayの作成

```python
# 基本的な作成方法
list1 = [4, 5, 6] 
arr = np.array(list1)
print(arr.ndim)
print(arr)
arr = np.array(list1,dtype=np.float64)
print(arr)
```

```python
# コピー
arr1 = np.copy(arr)
print(arr1)
```
